```python
def get_dataset_stats(values):
  return min(values), max(values), sum(values) / len(values)
# 강의 : 함수에서 여러개의 값을 리턴하면 이를 각각 원소로 하는 튜플로 넘어옴 (아래 minimum ~ avarage 참고)

data = [12, 8, 21, 17, 5, 32, 14]

# 함수값 return 언패킹
minimum, maximum, average = get_dataset_stats(data) ## 강의
print(f'최솟값: {minimum}, 최댓값: {maximum}, 평균: {average:.2f}')
```

```python
student_scores = [
    ('김철수', 85, 92, 78),
    ('이영희', 92, 88, 95),
    ('박지민', 75, 83, 90)
]

# 반복문에서 언패킹 사용
for name, database, python, cloud in student_scores:   # 강의 : for number in student_scores:로 하는 경우 number가 튜플이 됨. 이 경우 name, database, python, cloud로 쪼개서 각각의 값을 분해해서 받게 됨
  average = (database + python + cloud) / 3
  print(f'{name}의 평균 점수: {average:.1f}')
```

```python
monthly_sales = [1200, 1350, 1420, 1500, 1300, 1580, 1620, 1700, 1800, 1850, 1900, 2000]

# 여러 값을 묶어서 언패킹
# 강의 : 1월, 2월 및 나머지 월 매출액
first, second, *remaining = monthly_sales  # *이용해서 나머지 모두.
print(f'1월 판매액: {first}')
print(f'2월 판매액: {second}')
print(f'나머지 월 판매액: {remaining}')

# 강의 : 1월, 12월 및 나머지 월 매출액
first, *middle, last = monthly_sales #중간에 *도 가능

print(f'첫 달(1월) 판매액: {first}')
print(f'마지막 달(12월) 판매액: {last}')
print(f'중간 달(2~11월)의 판매액: {middle}')
```

```python
number = int(input('정수를 입력하세요: '))

# 강의 : 예외가 발생할 수도 있는 부분에 try 넣고 들여쓰기 옮겨

# try 문
try:
    result = 10 / number   # 강의 : number에 0 넣어서 오류난 경우 12번으로 이동
    print(f'결과: {result}')

# except 문
except:  # 강의
    print('0으로 나눌 수 없습니다.')
```

```python
## 강의 : 예외가 여러 개인 경우

# try 문
try:
    number = int(input('정수를 입력하세요: '))   # 원래처럼 맨 위에 위치시 try가 캐치를 못함. 원래 위치에 있으면 ValueError는 절대 발생하지 않음
    result = 10 / number   # 강의 : number에 0 넣어서 오류난 경우 12번으로 이동
    print(f'결과: {result}')

# except 문
except ZeroDivisionError:
    print('0으로 나눌 수 없습니다.')
except ValueError:
    print('유효한 숫자를 입력해야합니다')
```
